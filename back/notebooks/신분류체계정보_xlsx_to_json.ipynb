```python
import pandas as pd
import json

file_name = "./data/신분류체계정보_관광타입정보_연계_정의서.xlsx"
df = pd.read_excel(file_name, header=[3, 4])

top_headers = pd.Series([col[0] for col in df.columns])
top_headers = top_headers.mask(top_headers.str.contains('Unnamed')).ffill().fillna("")

new_columns = []
for col in df.columns:
    top = col[0] if "Unnamed" not in col[0] else ""
    bottom = col[1]
    
    if top:
        top_clean = top.split("(")[0]
        new_columns.append(f"{top_clean}_{bottom}")
    else:
        new_columns.append(bottom)

df.columns = new_columns

df['대분류_대분류코드'] = df['대분류_대분류코드'].ffill()
df['대분류_대분류명'] = df['대분류_대분류명'].ffill()
df['중분류_중분류코드'] = df['중분류_중분류코드'].ffill()
df['중분류_중분류명'] = df['중분류_중분류명'].ffill()

df = df.dropna(subset=['소분류_소분류코드'])

print("새로 정의된 컬럼명:", df.columns.tolist())

df.head(10)
```

```python
category_dict = {}

for index, row in df.iterrows():
    key = str(row['소분류_소분류코드']).strip()
    
    if key.lower() == 'nan' or not key:
        continue
        
    category_dict[key] = {
        "대분류": str(row['대분류_대분류명']).strip(),
        "중분류": str(row['중분류_중분류명']).strip(),
        "소분류명": str(row['소분류_소분류명']).strip(),
        "관광타입명": str(row['관광타입_관광타입명']).strip()
    }

# 저장하기
output_file = "./data/신분류체계정보_관광타입정보_연계_정의서.json"
with open(output_file, "w", encoding="utf-8") as f:
    json.dump(category_dict, f, ensure_ascii=False, indent=2)

print(f"총 {len(category_dict)}개의 카테고리가 {output_file}에 저장되었습니다.")
```
